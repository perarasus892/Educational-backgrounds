```python
# Drop rows where 'skills' is NaN before processing
df_skills = df.dropna(subset=['skills']).copy()

# Function to split skills and flatten the list
def get_all_skills(df_column):
    all_skills = []
    for skills_str in df_column:
        # Ensure skills_str is treated as string, then split and strip whitespace
        skills_list = [s.strip() for s in str(skills_str).split(',')]
        all_skills.extend(skills_list)
    return pd.Series(all_skills)

# Get all individual skills and count their occurrences
individual_skills = get_all_skills(df_skills['skills'])
skill_counts = individual_skills.value_counts()

# Display the top 10 most frequent skills
print("Top 10 most frequent skills:")
print(skill_counts.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the top 10 most frequent skills
plt.figure(figsize=(12, 7))
sns.barplot(x=skill_counts.head(10).index, y=skill_counts.head(10).values, palette='viridis')
plt.title('Top 10 Most Frequent Skills')
plt.xlabel('Skill')
plt.ylabel('Number of Occurrences')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# clean_and_prepare_education_job_data.py
import pandas as pd
import numpy as np
import re

IN_PATH = "/content/education_job_role.csv"
CLEAN_OUT = "/content/education_job_clean.csv"
ML_OUT = "/content/education_job_ml_ready.csv"

def simplify_qualification(q):
    if pd.isna(q):
        return np.nan
    q = str(q).lower()
    if "phd" in q or "doctor" in q:
        return "PhD"
    if "master" in q or "m.sc" in q or "mtech" in q or "mca" in q or "mba" in q:
        return "Master"
    if "bachelor" in q or "b.sc" in q or "btech" in q or "bca" in q or "bcom" in q:
        return "Bachelor"
    return "Other"

def map_experience_level(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()
    if s.startswith("entry") or s in ["fresh","junior","trainee"]:
        return "Entry"
    if s.startswith("mid") or s in ["intermediate"]:
        return "Mid"
    if s.startswith("senior") or s in ["lead","sr"]:
        return "Senior"
    return "Other"

def normalize_job_role(j):
    if pd.isna(j):
        return np.nan
    j = str(j).strip()
    # basic normalization: remove extra spaces, unify common variants
    j = re.sub(r"\s+", " ", j)
    # common fixes (add more rules if needed)
    j = j.replace("Frontend Developer", "Frontend Developer")
    j = j.replace("Web Developer", "Web Developer")
    return j

def skills_count(s):
    if pd.isna(s):
        return 0
    # split on comma/semicolon/pipe
    parts = re.split(r"[;,|/]+", str(s))
    parts = [p.strip() for p in parts if p.strip()]
    return len(parts)

def programming_skills_count(s):
    if pd.isna(s):
        return 0
    prog = {"python","java","c++","c#","c","javascript","sql","react","angular","node","django","flask","spring","php","ruby","tensorflow","keras"}
    parts = [p.strip().lower() for p in re.split(r"[;,|/]+", str(s)) if p.strip()]
    return sum(1 for p in parts if any(tok in p for tok in prog))

# --- Load
df = pd.read_csv(IN_PATH)

# --- Basic text cleaning
text_cols = ["skills", "qualification", "experience_level", "job_role", "Internship_Experience", "College_ID", "Placement"]
for c in text_cols:
    if c in df.columns:
        df[c] = df[c].astype(object).where(~df[c].isna(), None)  # keep NaN/None consistent
        df[c] = df[c].apply(lambda v: v.strip() if isinstance(v, str) else v)

# --- Remove exact duplicates (keep first)
df = df.drop_duplicates().reset_index(drop=True)

# --- Column-wise cleaning and feature extraction
if "qualification" in df.columns:
    df["qualification_simple"] = df["qualification"].apply(simplify_qualification)
else:
    df["qualification_simple"] = np.nan

if "experience_level" in df.columns:
    df["experience_level_clean"] = df["experience_level"].apply(map_experience_level)
else:
    df["experience_level_clean"] = np.nan

if "job_role" in df.columns:
    df["job_role_clean"] = df["job_role"].apply(normalize_job_role)
else:
    df["job_role_clean"] = np.nan

if "skills" in df.columns:
    df["skills_count"] = df["skills"].apply(skills_count)
    df["programming_skills_count"] = df["skills"].apply(programming_skills_count)
else:
    df["skills_count"] = 0
    df["programming_skills_count"] = 0

# --- Numeric columns: fill with median if present
numeric_cols = ["IQ", "Prev_Sem_Result", "CGPA", "Academic_Performance",
                "Extra_Curricular_Score", "Communication_Skills", "Projects_Completed"]
for col in numeric_cols:
    if col in df.columns:
        median = pd.to_numeric(df[col], errors="coerce").median()
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(median)

# --- Internship_Experience & Placement fill
if "Internship_Experience" in df.columns:
    df["Internship_Experience"] = df["Internship_Experience"].fillna("No")
if "Placement" in df.columns:
    df["Placement"] = df["Placement"].fillna("Unknown")

# --- Fill remaining textual NaNs with explicit marker (optional)
for c in df.select_dtypes(include=['object']).columns:
    df[c] = df[c].where(~df[c].isna(), "Unknown")

# --- Save basic cleaned CSV
df.to_csv(CLEAN_OUT, index=False)

# --- Create ML-ready version: drop low-value raw text, keep engineered features
ml_cols_keep = [
    "candidate_id", "College_ID",
    "qualification_simple", "experience_level_clean", "skills_count", "programming_skills_count",
    "IQ", "Prev_Sem_Result", "CGPA", "Academic_Performance", "Extra_Curricular_Score",
    "Communication_Skills", "Projects_Completed", "Internship_Experience", "Placement", "job_role_clean"
]
ml_df = df[[c for c in ml_cols_keep if c in df.columns]].copy()

# Rename target column to 'job_role' if present
if "job_role_clean" in ml_df.columns:
    ml_df = ml_df.rename(columns={"job_role_clean": "job_role"})

# Encode simple categorical columns (example: experience -> numeric)
exp_map = {"Entry":0, "Mid":1, "Senior":2, "Other":3, "Unknown":3}
if "experience_level_clean" in ml_df.columns:
    ml_df["experience_level_encoded"] = ml_df["experience_level_clean"].map(exp_map).fillna(3).astype(int)

# Save ML-ready CSV
ml_df.to_csv(ML_OUT, index=False)

print("Saved cleaned dataset ->", CLEAN_OUT)
print("Saved ML-ready dataset ->", ML_OUT)
```
